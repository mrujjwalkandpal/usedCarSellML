# Data Cleaning

### 1. Importing Modules

```python
import numpy as np
import pandas as pd
```

### 2. Loading Data

```python
df=pd.read_csv("../data/raw/car_details.xls")
log=[("raw data", len(df))]
```

```python
df_clean = df.copy()
```

### 3. Checking Missing Values

```python
df_clean.isnull().sum()
```

### 4. Checking Duplicates

```python
print("Total rows:", len(df_clean))
print("Duplicate rows:", df_clean.duplicated().sum())
```

##### a. Inspecting the Rows

```python
duplicates = df_clean[df_clean.duplicated(keep=False)]

duplicates.sort_values(by=['name','year']).head(15)
```

#### b. Removing Duplicated

```python
df_clean = df_clean.drop_duplicates()
print(
    "Rows after removing duplicates:",
    len(df_clean)
)
```

> Duplicate Removal: The dataset contained 763 duplicate occurrences across 4,340 records. Exact duplicate rows were removed using drop_duplicates(). The cleaned dataset now contains 3,577 records.

#### c. Verifying Duplicated

```python
print("Duplicates Remains : ", df_clean.duplicated().sum())
```

### 5. Inspecting Numerical Data

```python
df_clean[['year','selling_price','km_driven']].describe()
```

#### **Find suspicious records**

##### *A. Cars with extremely high mileage*

```python
df_clean[df_clean['km_driven']>300000]
```

##### *B. Cars with unusually high selling prices*

```python
df_clean[df_clean['selling_price'] > 5000000]
```

##### *C. Cars with very low mileage*

```python
df_clean[df_clean['km_driven'] < 100]
```

##### *D. Check logical validity*

```python
print("Negative prices:",
      (df_clean['selling_price'] < 0).sum())

print("Zero prices:",
      (df_clean['selling_price'] == 0).sum())

print("Negative mileage:",
      (df_clean['km_driven'] < 0).sum())

print("Invalid future years:",
      (df_clean['year'] > 2026).sum())
```

```python
for col in ['fuel', 'seller_type', 'transmission', 'owner']:
    print(f"\n{col}:")
    print(df_clean[col].value_counts())
```

##### *E. Investigating Name Column*

```python
print("Unique car names:", df_clean['name'].nunique())

print("Total records:", len(df_clean))

print(df_clean['name'].value_counts().head(10))
```

### 6. Exporting Cleaned Data

```python
df_clean.to_csv("../data/processed/cleaned_car_details.xls",index=False)
```
